```python
# Memasukkan nilai Y dan X
Y = int(input("Masukkan nilai Y: "))
X = int(input("Masukkan nilai X: "))

# Algoritma Sixth Macro
TEMP = Y
Y = 1

while X > 0:
  X = X - 1
  Y = Y * TEMP

# Menampilkan hasil
print("Hasil perpangkatan adalah:", Y)
```
